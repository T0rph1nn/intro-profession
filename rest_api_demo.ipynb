# REST API demo: OMDb + .env

API-ключ читается из переменной окружения `OMDB_API_KEY`.

```python
import json
import os

import requests
from dotenv import load_dotenv
```

```python
load_dotenv()

title = input("Введите фильм для поиска: ")
api_key = os.getenv("OMDB_API_KEY")

if not api_key:
    raise ValueError("Ключ OMDB_API_KEY не найден в .env")
```

```python
response = requests.get(
    "http://www.omdbapi.com/",
    params={
        "t": title,
        "apikey": api_key,
    },
    timeout=10,
)
response.raise_for_status()
```

```python
json_data = response.json()
print(json.dumps(json_data, ensure_ascii=False, indent=2))
```

```python
movie_title = json_data.get("Title")
year = json_data.get("Year")
genre = json_data.get("Genre")
director = json_data.get("Director")
imdb_rating = json_data.get("imdbRating")
actors = json_data.get("Actors")

print(f"Название: {movie_title}")
print(f"Год: {year}")
print(f"Жанр: {genre}")
print(f"Режиссёр: {director}")
print(f"Актёры: {actors}")
print(f"Рейтинг IMDb: {imdb_rating}")
print(f"Фильм '{movie_title}' ({year}) — жанр: {genre}. Режиссёр: {director}.")
```
