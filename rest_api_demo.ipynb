# REST API demo: OMDb + .env

Версия после code review: код оформлен через функции и проверяет ответ OMDb.

```python
import json
import os

import requests
from dotenv import load_dotenv
```

```python
API_URL = "http://www.omdbapi.com/"

def get_api_key() -> str:
    load_dotenv()
    api_key = os.getenv("OMDB_API_KEY")
    if not api_key:
        raise ValueError("Ключ OMDB_API_KEY не найден в .env")
    return api_key

def fetch_movie(title: str, api_key: str) -> dict:
    response = requests.get(
        API_URL,
        params={"t": title, "apikey": api_key},
        timeout=10,
    )
    response.raise_for_status()
    return response.json()

def print_movie_info(json_data: dict) -> None:
    if json_data.get("Response") == "False":
        print("Фильм не найден:", json_data.get("Error"))
        return

    print(json.dumps(json_data, ensure_ascii=False, indent=2))
    movie_title = json_data.get("Title", "Неизвестно")
    year = json_data.get("Year", "Неизвестно")
    genre = json_data.get("Genre", "Неизвестно")
    director = json_data.get("Director", "Неизвестно")
    actors = json_data.get("Actors", "Неизвестно")
    imdb_rating = json_data.get("imdbRating", "Нет рейтинга")

    print(f"Название: {movie_title}")
    print(f"Год: {year}")
    print(f"Жанр: {genre}")
    print(f"Режиссёр: {director}")
    print(f"Актёры: {actors}")
    print(f"Рейтинг IMDb: {imdb_rating}")
    print(f"Фильм '{movie_title}' ({year}) — жанр: {genre}. Режиссёр: {director}.")
```

```python
title = input("Введите фильм для поиска: ").strip()

if not title:
    raise ValueError("Название фильма не должно быть пустым")

movie_data = fetch_movie(title, get_api_key())
print_movie_info(movie_data)
```
